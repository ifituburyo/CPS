### 1. Autoencoder-Based Anomaly Detection (LSTM Autoencoder)


Train on normal segments (e.g., 0–6 hrs, 11–end).

Reconstruct input using LSTM encoder-decoder.

Compute reconstruction error at each time step.

High error during 6–11 hrs = potential leak.

### 2. Statistical Models (e.g., ARIMA, Moving Average Residuals)
#### Why: If the signal is mostly smooth or seasonal, these models can detect deviations.

###### How it works:

Fit ARIMA model on clean part.

Forecast next steps.

Compare forecast vs actual.

Large residuals in leak window.
